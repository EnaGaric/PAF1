```python
import matplotlib.pyplot as plt

print("Unesi koordinate za dvije točke:\n")

# Funkcija za unos koordinata s provjerom
def unos_koordinata():
    while True:
        try:
            x1 = float(input("x1: "))
            y1 = float(input("y1: "))
            x2 = float(input("x2: "))
            y2 = float(input("y2: "))
            return x1, y1, x2, y2
        except ValueError:
            print("Molim te da uneseš broj u decimalnom zapisu.\n")

x1, y1, x2, y2 = unos_koordinata()

#pravac moze biti vertikalan
if x1 == x2:
    print(f"Pravac je vertikalan: x = {x1}")
    vertikalan = True
else:
    k = (y2 - y1) / (x2 - x1)
    l = y1 - k * x1
    print("Jednadzba pravca:")
    print("y = {}x + {}".format(k, l))
    vertikalan = False


#crtanje
plt.xlabel('x')
plt.ylabel('y')
plt.legend
plt.grid(True)


if not vertikalan:
    plt.plot([x1, x2], [y1, y2], color='blue', label='Pravac')
else:
    plt.axvline(x=x1, color='blue', label='Pravac')


# opcija prikaza ili spremanja
opcija = input("Prikazati graf ili spremiti graf kao PDF? Unesi sljedeće: (ekran/pdf): ").strip().lower()
if opcija == 'ekran':
    plt.show()
elif opcija == 'pdf':
    ime = input("Unesi ime PDF datoteke (bez .pdf): ").strip() + '.pdf'
    plt.savefig(ime)
    print(f"Graf spremljen kao {ime}")
```
